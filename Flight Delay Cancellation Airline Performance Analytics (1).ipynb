* **CANCELLATION_CODE:** Code representing the reason for cancellation.

```python
# 1. Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display
from google.colab import files

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

```python
# 2. Upload the Raw Dataset

uploaded = files.upload()

print("Dataset upload completed.")
```

```python
# 3. Identify the Uploaded CSV File

csv_files = [
    file_name
    for file_name in uploaded.keys()
    if file_name.lower().endswith(".csv")
]

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV file found. Please upload T_ONTIME_REPORTING.csv."
    )

file_path = csv_files[0]

print("Selected file:", file_path)
```

```python
# 4. Load the Raw Dataset

df_raw = pd.read_csv(
    file_path,
    low_memory=False
)

print("Raw dataset loaded successfully.")
print("Raw dataset shape:", df_raw.shape)

display(df_raw.head())
```

```python
# 5. Create a Working Copy

df = df_raw.copy()

print("Working copy created successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# 6. Initial Dataset Inspection

print("Dataset Information")
print("=" * 60)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame(name="Data Type"))

print("\nFirst five records:")
display(df.head())

print("\nStatistical summary:")
display(df.describe(include="all").T)
```

```python
# 7. Standardize Column Names

df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)
)

print("Column names standardized successfully.")

print("\nUpdated column names:")
print(df.columns.tolist())
```

```python
# 8. Check Duplicate Records

duplicate_count = df.duplicated().sum()

duplicate_percentage = (
    duplicate_count / len(df)
) * 100

print("Number of duplicate records:", duplicate_count)

print(
    "Duplicate percentage:",
    round(duplicate_percentage, 2),
    "%"
)
```

```python
# 9. Remove Duplicate Records

rows_before_duplicates = len(df)

df = df.drop_duplicates().reset_index(drop=True)

rows_after_duplicates = len(df)

duplicates_removed = (
    rows_before_duplicates - rows_after_duplicates
)

print("Rows before duplicate removal:", rows_before_duplicates)
print("Rows after duplicate removal:", rows_after_duplicates)
print("Duplicate records removed:", duplicates_removed)
```

```python
# 10. Convert the Flight Date Column

if "FL_DATE" in df.columns:

    df["FL_DATE"] = pd.to_datetime(
        df["FL_DATE"],
        errors="coerce"
    )

    print("FL_DATE converted successfully.")
    print("FL_DATE data type:", df["FL_DATE"].dtype)
    print("Minimum flight date:", df["FL_DATE"].min())
    print("Maximum flight date:", df["FL_DATE"].max())

else:

    print("FL_DATE column was not found.")
```

```python
# 11. Check Invalid or Missing Dates

if "FL_DATE" in df.columns:

    invalid_date_count = df["FL_DATE"].isna().sum()

    print(
        "Invalid or missing date records:",
        invalid_date_count
    )

else:

    print("FL_DATE column was not found.")
```

```python
# 12. Remove Invalid Date Records

if "FL_DATE" in df.columns:

    rows_before_date_cleaning = len(df)

    invalid_date_count = df["FL_DATE"].isna().sum()

    df = df.dropna(
        subset=["FL_DATE"]
    ).reset_index(drop=True)

    rows_after_date_cleaning = len(df)

    print(
        "Invalid date records found:",
        invalid_date_count
    )

    print(
        "Invalid date records removed:",
        rows_before_date_cleaning - rows_after_date_cleaning
    )

    print(
        "Rows remaining after date cleaning:",
        rows_after_date_cleaning
    )

else:

    print("FL_DATE column was not found.")
```

```python
# 13. Convert Numeric Columns

numeric_columns = [
    "YEAR",
    "QUARTER",
    "MONTH",
    "DAY_OF_MONTH",
    "DAY_OF_WEEK",
    "DEP_DELAY",
    "DEP_DELAY_NEW",
    "DEP_DEL15",
    "ARR_DELAY",
    "ARR_DELAY_NEW",
    "ARR_DEL15",
    "AIR_TIME",
    "FLIGHTS",
    "DISTANCE",
    "CANCELLED"
]

existing_numeric_columns = [
    column
    for column in numeric_columns
    if column in df.columns
]

for column in existing_numeric_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print("Numeric columns converted successfully.")

display(
    df[existing_numeric_columns]
    .dtypes
    .to_frame(name="Data Type")
)
```

```python
# 14. Clean Text Columns

text_columns = [
    "OP_UNIQUE_CARRIER",
    "ORIGIN",
    "ORIGIN_CITY_NAME",
    "DEST",
    "DEST_CITY_NAME",
    "CANCELLATION_CODE"
]

existing_text_columns = [
    column
    for column in text_columns
    if column in df.columns
]

for column in existing_text_columns:

    df[column] = (
        df[column]
        .astype("string")
        .str.strip()
    )

print("Text columns cleaned successfully.")

display(
    df[existing_text_columns].head()
)
```

```python
# 15. Check Missing Values

missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
].sort_values(
    by="Missing Count",
    ascending=False
)

print("Missing value summary:")

display(missing_summary)
```

```python
# 16. Visualize Missing Values

if not missing_summary.empty:

    plt.figure(figsize=(12, 6))

    missing_summary["Missing Count"].plot(
        kind="bar"
    )

    plt.title("Missing Values by Column")
    plt.xlabel("Columns")
    plt.ylabel("Number of Missing Records")
    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()

else:

    print("No missing values found in the dataset.")
```

```python
# 17. Create Cancellation Flag

if "CANCELLED" in df.columns:

    df["CANCELLED_FLAG"] = (
        df["CANCELLED"]
        .fillna(0)
        .astype(int)
    )

else:

    df["CANCELLED_FLAG"] = 0

print("Cancellation flag created successfully.")

display(
    df["CANCELLED_FLAG"]
    .value_counts(dropna=False)
    .to_frame(name="Flight Count")
)
```

```python
# 18. Handle Cancellation Code

if "CANCELLATION_CODE" in df.columns:

    df["CANCELLATION_CODE"] = (
        df["CANCELLATION_CODE"]
        .fillna("Not Cancelled")
    )

    print("Cancellation codes handled successfully.")

    display(
        df["CANCELLATION_CODE"]
        .value_counts(dropna=False)
        .to_frame(name="Flight Count")
    )

else:

    print("CANCELLATION_CODE column was not found.")
```

```python
# 19. Create Departure Delay Flag

if "DEP_DELAY" in df.columns:

    df["DEP_DELAY_FLAG"] = np.where(
        df["DEP_DELAY"] > 15,
        1,
        0
    )

    print("Departure delay flag created successfully.")

else:

    print("DEP_DELAY column was not found.")
```

```python
# 20. Create Arrival Delay Flag

if "ARR_DELAY" in df.columns:

    df["ARR_DELAY_FLAG"] = np.where(
        df["ARR_DELAY"] > 15,
        1,
        0
    )

    print("Arrival delay flag created successfully.")

else:

    print("ARR_DELAY column was not found.")
```

```python
# 21. Create Departure Delay Status

if "DEP_DELAY" in df.columns:

    df["DEP_DELAY_STATUS"] = np.select(
        [
            df["DEP_DELAY"].isna(),
            df["DEP_DELAY"] <= 0,
            df["DEP_DELAY"] <= 15,
            df["DEP_DELAY"] <= 60,
            df["DEP_DELAY"] > 60
        ],
        [
            "Not Available",
            "On Time / Early",
            "Minor Delay",
            "Moderate Delay",
            "Severe Delay"
        ],
        default="Not Available"
    )

    print("Departure delay status created successfully.")

    display(
        df[
            [
                "DEP_DELAY",
                "DEP_DELAY_STATUS"
            ]
        ].head()
    )

else:

    print("DEP_DELAY column was not found.")
```

```python
# 22. Create Arrival Delay Status

if "ARR_DELAY" in df.columns:

    df["ARR_DELAY_STATUS"] = np.select(
        [
            df["ARR_DELAY"].isna(),
            df["ARR_DELAY"] <= 0,
            df["ARR_DELAY"] <= 15,
            df["ARR_DELAY"] <= 60,
            df["ARR_DELAY"] > 60
        ],
        [
            "Not Available",
            "On Time / Early",
            "Minor Delay",
            "Moderate Delay",
            "Severe Delay"
        ],
        default="Not Available"
    )

    print("Arrival delay status created successfully.")

    display(
        df[
            [
                "ARR_DELAY",
                "ARR_DELAY_STATUS"
            ]
        ].head()
    )

else:

    print("ARR_DELAY column was not found.")
```

```python
# 23. Create Date-Based Features

if "FL_DATE" in df.columns:

    df["YEAR_CALCULATED"] = df["FL_DATE"].dt.year

    df["MONTH_NAME"] = (
        df["FL_DATE"].dt.month_name()
    )

    df["DAY_NAME"] = (
        df["FL_DATE"].dt.day_name()
    )

    df["WEEK_OF_YEAR"] = (
        df["FL_DATE"]
        .dt.isocalendar()
        .week
        .astype("Int64")
    )

    df["QUARTER_CALCULATED"] = (
        "Q" + df["FL_DATE"]
        .dt.quarter
        .astype("string")
    )

    print("Date-based features created successfully.")

    display(
        df[
            [
                "FL_DATE",
                "YEAR_CALCULATED",
                "MONTH_NAME",
                "DAY_NAME",
                "WEEK_OF_YEAR",
                "QUARTER_CALCULATED"
            ]
        ].head()
    )

else:

    print("FL_DATE column was not found.")
```

```python
# 24. Create Flight Duration Category

if "AIR_TIME" in df.columns:

    df["FLIGHT_DURATION_CATEGORY"] = pd.cut(
        df["AIR_TIME"],
        bins=[
            -np.inf,
            60,
            180,
            360,
            np.inf
        ],
        labels=[
            "Short Haul",
            "Medium Haul",
            "Long Haul",
            "Very Long Haul"
        ]
    )

    print("Flight duration category created successfully.")

    display(
        df[
            [
                "AIR_TIME",
                "FLIGHT_DURATION_CATEGORY"
            ]
        ].head()
    )

else:

    print("AIR_TIME column was not found.")
```

```python
# 25. Create Distance Category

if "DISTANCE" in df.columns:

    df["DISTANCE_CATEGORY"] = pd.cut(
        df["DISTANCE"],
        bins=[
            -np.inf,
            500,
            1500,
            3000,
            np.inf
        ],
        labels=[
            "Short Distance",
            "Medium Distance",
            "Long Distance",
            "Very Long Distance"
        ]
    )

    print("Distance category created successfully.")

    display(
        df[
            [
                "DISTANCE",
                "DISTANCE_CATEGORY"
            ]
        ].head()
    )

else:

    print("DISTANCE column was not found.")
```

```python
# 26. Validate Important Columns

validation_results = {}

if "MONTH" in df.columns:

    validation_results["Invalid Month Values"] = (
        ~df["MONTH"].between(1, 12)
    ).sum()

if "QUARTER" in df.columns:

    validation_results["Invalid Quarter Values"] = (
        ~df["QUARTER"].between(1, 4)
    ).sum()

if "DAY_OF_WEEK" in df.columns:

    validation_results["Invalid Day of Week Values"] = (
        ~df["DAY_OF_WEEK"].between(1, 7)
    ).sum()

if "CANCELLED_FLAG" in df.columns:

    validation_results["Invalid Cancellation Values"] = (
        ~df["CANCELLED_FLAG"].isin([0, 1])
    ).sum()

validation_report = pd.DataFrame(
    validation_results.items(),
    columns=[
        "Validation Check",
        "Invalid Record Count"
    ]
)

display(validation_report)
```

```python
# 27. Check Negative and Invalid Values

check_columns = [
    column
    for column in [
        "DEP_DELAY",
        "ARR_DELAY",
        "AIR_TIME",
        "DISTANCE"
    ]
    if column in df.columns
]

print("Descriptive statistics for important numeric columns:")

display(
    df[check_columns].describe().T
)

if "AIR_TIME" in df.columns:

    print(
        "AIR_TIME values less than or equal to zero:",
        (df["AIR_TIME"] <= 0).sum()
    )

if "DISTANCE" in df.columns:

    print(
        "DISTANCE values less than or equal to zero:",
        (df["DISTANCE"] <= 0).sum()
    )
```

```python
# 28. Create Data Quality Report

data_quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Missing Count": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2),
    "Unique Values": df.nunique().values
})

print("Data quality report created successfully.")

display(data_quality_report)
```

```python
# 29. Export Data Quality Report

data_quality_report.to_csv(
    "flight_data_quality_report.csv",
    index=False
)

print(
    "Data quality report saved as:",
    "flight_data_quality_report.csv"
)
```

```python
# 30. Display Final Cleaned Dataset

print("Final cleaned dataset preview:")

display(df.head())

print("\nFinal dataset shape:", df.shape)

print("\nFinal data types:")
display(df.dtypes.to_frame(name="Data Type"))
```

```python
# 31. Final Data Cleaning Summary

print("FINAL DATA CLEANING SUMMARY")
print("=" * 60)

print("Raw dataset rows:", len(df_raw))
print("Final cleaned rows:", len(df))
print("Rows removed:", len(df_raw) - len(df))
print("Final number of columns:", len(df.columns))
print("Duplicate rows remaining:", df.duplicated().sum())
print(
    "Total missing values remaining:",
    df.isnull().sum().sum()
)

if "FL_DATE" in df.columns:

    print(
        "Flight date range:",
        df["FL_DATE"].min(),
        "to",
        df["FL_DATE"].max()
    )
```

```python
# 32. Export the Cleaned Dataset

cleaned_file_name = (
    "flight_delay_cleaned_preprocessed.csv"
)

df.to_csv(
    cleaned_file_name,
    index=False
)

print(
    "Cleaned dataset exported successfully as:",
    cleaned_file_name
)
```

2. These extreme values may represent operational disruptions and should be considered when analyzing overall airline performance.

### 6. Average Departure Delay by Airline

**Business Question**:
Which airlines have the highest and lowest average departure delays?

### 14. Statistical Summary of Flight Performance

```python
summary_columns = [
    "DEP_DELAY",
    "ARR_DELAY",
    "AIR_TIME",
    "DISTANCE"
]

statistical_summary = df[summary_columns].describe().T

statistical_summary["Median"] = df[summary_columns].median()
statistical_summary["Variance"] = df[summary_columns].var()

display(statistical_summary.round(2))
```

### Interpretation:

The statistical summary provides an overview of departure delays, arrival delays, air time, and flight distance using measures such as mean, median, standard deviation, quartiles, and variance.

- **Departure Delay:** The average departure delay is 15.70 minutes, while the median is -2 minutes. This difference suggests that some large delays may be increasing the average.

- **Arrival Delay:** The average arrival delay is 10.35 minutes, while the median is -5 minutes. This indicates that the average is higher than the middle observation, suggesting a right-skewed distribution.

- **Air Time:** The average air time is 116.34 minutes, and the median is 100 minutes. The maximum recorded air time is 723 minutes.

- **Flight Distance:** The average flight distance is 836.35 miles, with a median of 693 miles. The maximum recorded distance is 5,095 miles.

- **Variability:** Departure and arrival delays have high standard deviations of 64.18 and 66.78 minutes, respectively, indicating considerable variation in delay values.

### Key Observation:

1. The mean departure and arrival delays are greater than their respective medians.

2. The large maximum delay values also suggest the presence of extreme observations that may influence the averages.
